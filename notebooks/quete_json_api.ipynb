````python
with open("../data/villes.json", encoding="utf-8") as v:
    data2 = json.load(v)
````

Étape 1 — explorer sans pandas

````python
#le nombre de villes
print(len(data2['villes']))

#le nom de la deuxième ville
print(data2['villes'][1]['nom'])

#le nombre de restaurants à Syracuse
print(len(data2['villes'][2]['restaurants']))

#le nom du premier restaurant de Palerme
print(data2['villes'][1]['restaurants'][0]['nom'])

````

Étape 2 — compter à la main

````python
for ville in data2['villes']:
    print(ville['nom'], len(ville['restaurants']))
````

Étape 3 — tout mettre à plat

````python
df2 = pd.json_normalize(data2, record_path=["villes", "restaurants"], meta=[["villes", "nom"], ["villes", "region"], ["villes", "habitants"],],)

df2_intermediaire = df2[["nom", "type", "notes", "villes.region"]]

print(df2_intermediaire.shape)  # (7, 4)
print(df2_intermediaire.columns.tolist())
df2_intermediaire
````

Étape 4 — exploiter

````python
moyenne = []

for note in df2['notes']:
    resultat = sum(note) / len(note)
    moyenne.append(round(resultat, 2))

df2['note_moyenne'] = moyenne

# Question 1 : moyenne des restaurants de chaque ville
moyennes_villes = df2.groupby("villes.nom")["note_moyenne"].mean()
print(moyennes_villes.sort_values(ascending=False).round(2))

# Question 2 : nombre de restaurants de chaque type
print(df2["type"].value_counts())

# Afficher le tableau final
df2


    



````
